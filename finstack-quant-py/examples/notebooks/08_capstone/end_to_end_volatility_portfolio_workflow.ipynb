### Track-specific requirements

Credit submissions must explain the split-PIK callable bond, revolver's contingent exposure, convertible debt/equity interaction, and credit-index option's front-end protection. The BBB sleeve must retain actual collateral and waterfall assumptions, with native-book, BBB and combined subtotals on every relevant date and stress. The credit reference prices all three tranches under stress, solves the BBB first-writedown threshold, traces a controlled one-period stressed cash distribution, reports unconditional versus default-conditional expected revolver draws, and reprices convertible parity/floor/delta across stock regimes. The draw model uses stated synthetic Bernoulli probabilities; it does not claim the observed 40% utilization is a conditional EAD forecast. An index hedge must disclose the difference between quote-space index CS01 and tranche z-spread CS01.

Volatility submissions must identify delivery-bond/CTD and basis risk, variance notional and realized/implied blending, commodity contract units and spread correlation. The UST future's reset mark plus cumulative variation cash must equal its quote-to-entry wealth representation. The volatility reference also reports fixed option-inventory replication P&L with its forward/cash correction, realizes squared-return versus remaining-implied-variance attribution, and compares delivery candidates and gross basis. The stressed finite-strike replication residual is about USD 62 and must remain visible; the acceptance bound is USD 1,000 on a USD 1 million variance notional. The native held future uses a fixed-deliverable proxy, so the changed candidate CTD is a reported model/basis risk rather than an automatic delivery-option valuation. Physical future theta is not available from the cashflow-theta path; the reference reports this limit rather than demanding a nonexistent metric.

### PM recommendation in each reference

The credit tear sheet proposes a conditional 25% reduction of the USD 15 million-face BBB holding: USD 3.75 million face. At the stated model marks and before costs, the sale proceeds held as USD cash preserve starting wealth and reduce the composed-stress loss by about USD 2.77 million. The volatility tear sheet proposes reducing the long Treasury future from ten to five contracts while delivery-basket basis is reviewed; the stated stress loss falls by about USD 51,158. These are sized responses to the displayed scenarios, not estimates of expected return or claims that the model marks are executable. Both references set additional borrower-loan exposure to zero pending the four-quarter DSCR remediation, without treating existing revolver commitments as cancellable. The reports state the monitoring triggers and retained risks beside each recommendation.

### Scope of the executable references

The references use synthetic shared inputs and actual native valuations. They independently reprice each native position and separately value credit BBB. The composed stress changes USD discount calibration quotes while holding the SOFR projection basis fixed; its IRS result is not a parallel bump to every rate curve. Stress recalibrates relevant credit quotes with the shocked discount environment; PV-only aggregation avoids accidentally asking for a quote-space risk replay under inconsistent fixed-hazard inputs. The combined market shock is an explicit scenario, not a forecast probability. Tail risk separately reprices the actual holdings under 504 seeded synthetic one-day shocks across eight stated factors, including BBB under a disclosed collateral-default mapping. Native VaR/ES are checked against explicit sample quantile/tail calculations. Factor loadings are fitted to those repriced returns; covariance contributions plus nonlinear residual variance reconcile to total variance. Bootstrap intervals describe finite-sample uncertainty under this scenario model, not uncertainty about whether the model is correct. The factors omit tenor shape, issuer/subcurve basis, smile, correlation and liquidity shocks; this is not a complete regulatory risk model.

Margin covers the actual common USD payer IRS as an eligible bilateral OTC example. Its remaining maturity comes from the contract's 15 April 2028 end date. Schedule IM and sensitivity-based SIMM are both computed, and the reused one-factor SIMM arithmetic check is retained. The native IR input representation nets tenors and does not implement separate subcurve basis aggregation. The result is therefore a supported IR illustration, not a complete all-risk-class or basis-complete SIMM submission. It excludes credit, equity, variance and commodity positions from the margin set. Exchange initial margin remains a qualitative external methodology in V1.

The regulatory-CSA VM example has zero threshold/independent amount and USD 500,000 MTA. Both actual IRS exposures are below that threshold for transfer, so no collateral moves. Zero VM delivery is checked against those terms; it is not zero exposure or zero risk. IM, VM, SA-CCR capital and funding cost are different quantities, and this reference does not infer capital from its partial IM amount.

The borrower tables contain a forecast and current-quarter EBITDA annualization, not a historical trailing-twelve-month claim. They support the held USD 1 million loan while the issuer's full debt balance remains a separate entity-level measure. The base DSCR fails all four forecast tests. Physical ECL uses the shared annual conditional PD and LGD, a disclosed watchlist/SICR policy, 7.5% EIR and the held loan's quarterly amortizing EAD. Native Stage-2 ECL reconciles to the visible bucket sum and is reported beside market PV, not deducted again. A pricing discount spread is not automatically a real-world default probability.

Historical-pattern stress separately replays the synthetic 2025Q2 observation-to-observation rate, credit, equity and FX changes onto the frozen 15 January book. Those illustrative observation dates follow the held date, so this is a scenario pattern rather than a contemporaneous backtest or evidence from a real crisis. Volatility, energy, SOFR projection basis and BBB collateral assumptions stay fixed in that replay; the separate composed/default stress covers other moves. Current-date curves and per-position/native/BBB totals reconcile before the replay is reported.

The reference performance window has four weekly valuation dates from 15 January to 5 February 2025, producing three independent periods. It deliberately precedes contractual distributions and has no external trades or flows. Each date uses contemporary FX; credit BBB is repriced at that date, and volatility variance receives only realized observations available by that date. This shorter capstone window is separate from Part III's three-quarter, flow-inclusive history. A student choosing a longer window must include the actual flows and futures settlement cash, rather than applying the no-flow shortcut beyond its domain.

## 2. Pricing — reconcile every position and the BBB sleeve

The native portfolio reports each position in native and base currency. A StructuredCredit envelope values all tranches, so it must not represent a BBB position. Price BBB through its tranche API and add that one claim to the report exactly once.

```python
base=value_portfolio(book,market,metrics=[])
wire=json.loads(base.to_json())
positions=pd.DataFrame([{"position":iid,"currency":row["value_native"]["currency"],
    "native_value":float(row["value_native"]["amount"]),"usd_value":float(row["value_base"]["amount"])}
    for iid,row in wire["position_values"].items()])
assert abs(positions.usd_value.sum()-base.total_value)<.01
bbb=structured_credit_tranche_metrics(json.dumps(holding["deal"]),"BBB",market,AS_OF) if holding else None
bbb_pv=holding["quantity"]*bbb.pv if holding else 0.0
combined_base=base.total_value+bbb_pv
if holding:
    positions.loc[len(positions)]={"position":"CLO-BBB","currency":"USD","native_value":bbb_pv,"usd_value":bbb_pv}
assert abs(positions.usd_value.sum()-combined_base)<.01
print(positions)
```

```python
from _shared.instrument_fixtures import instrument_envelope
individual_rows=[]
for position in book_input["positions"]:
    one=price_instrument(json.dumps(instrument_envelope(position["instrument_spec"])),market,AS_OF,metrics=[])
    currency=str(one.value.currency)
    fx_rate=1.0 if currency=="USD" else market.fx_required().rate(currency,"USD",AS_OF).rate
    direct_native=one.value.amount*position["quantity"]
    portfolio_native=float(wire["position_values"][position["instrument_id"]]["value_native"]["amount"])
    assert abs(direct_native-portfolio_native)<.01
    individual_rows.append({"position":position["instrument_id"],"currency":currency,
                            "direct_native":direct_native,"direct_USD":direct_native*fx_rate})
individual_prices=pd.DataFrame(individual_rows)
direct_total=individual_prices.direct_USD.sum()+bbb_pv
assert abs(direct_total-combined_base)<.01
print(individual_prices)
print({"independent_individual_plus_BBB":direct_total,"native_portfolio_plus_BBB":combined_base})
```

```python
# Independent closed-form calculation for the actual common European equity call.
equity=tracks.common.instruments("common")["SPX-CALL"]["instrument"]["spec"]
tau=(date.fromisoformat(equity["expiry"])-AS_OF).days/365
spot=float(market.get_price(equity["spot_id"])[0]);q=float(market.get_price(equity["div_yield_id"])[0])
K=float(equity["strike"]);sigma=market.get_surface(equity["vol_surface_id"]).vol(tau,K)
discount=market.get_discount(equity["discount_curve_id"]).df(tau);r=-math.log(discount)/tau
d1=(math.log(spot/K)+(r-q+.5*sigma*sigma)*tau)/(sigma*math.sqrt(tau));d2=d1-sigma*math.sqrt(tau)
N1=.5*(1+math.erf(d1/math.sqrt(2)));N2=.5*(1+math.erf(d2/math.sqrt(2)))
analytical_call=(spot*math.exp(-q*tau)*N1-K*discount*N2)*float(equity["notional"]["amount"])
native_call=float(wire["position_values"]["SPX-CALL"]["value_native"]["amount"])
analytical_price_error=analytical_call-native_call
assert abs(analytical_price_error)<.01
print({"independent_BSM_USD":analytical_call,"native_equity_call_USD":native_call,"difference_USD":analytical_price_error})
```

```python
# A scenario distribution, not observed historical returns or a calibrated forecast.
# Reprice the actual held book at the fixed valuation date; do not proxy it with the
# differently weighted statistical portfolio from lesson 3.2.
from finstack_quant.core.market_data import MarketContext
from finstack_quant.analytics import Performance

factor_names=["USD_zero_and_forward_decimal","EUR_zero_decimal","hazard_log_change",
              "equity_log_return","absolute_volatility_change","EURUSD_log_return",
              "energy_log_return","WTI_basis_log_return"]
factor_sd=np.array([.0005,.0004,.06,.012,.008,.006,.018,.008])
correlation=np.eye(8)
for i,j,rho in [(0,1,.30),(2,3,-.25),(3,4,-.40),(3,6,.20),(6,7,.20)]:
    correlation[i,j]=correlation[j,i]=rho
assert np.linalg.eigvalsh(correlation).min()>0
rng=np.random.default_rng(2025011508)
factor_panel=rng.standard_normal((504,8))@np.linalg.cholesky(correlation).T*factor_sd
frozen_market=json.loads(market.to_json())

def capstone_scenario_value(shock):
    # Each mutation starts from the same snapshot. Rates are continuous zero/forward
    # changes, hazards are multiplicative intensity changes, and all PV calls are
    # explicitly PV-only: this is curve space, not quote-recalibrated CS01.
    usd,eur,hazard,equity,volatility,fx_move,energy,basis=shock
    snapshot=deepcopy(frozen_market)
    for curve in snapshot["curves"]:
        if curve["type"]=="discount":
            shift=eur if curve["id"]=="EUR-OIS" else usd
            if curve["id"]=="CONVERT-RISKY":shift+=.025*math.expm1(hazard)
            curve["knot_points"]=[[t,df*math.exp(-shift*t)] for t,df in curve["knot_points"]]
            curve.pop("rate_calibration",None)
        elif curve["type"]=="forward":
            curve["knot_points"]=[[t,rate+usd] for t,rate in curve["knot_points"]]
        elif curve["type"]=="hazard":
            curve["knot_points"]=[[t,intensity*math.exp(hazard)] for t,intensity in curve["knot_points"]]
            curve.pop("hazard_calibration",None);curve["par_points"]=[]
        elif curve["type"]=="price":
            log_move=energy+(basis if curve["id"]=="WTI-FORWARD" else 0.)
            curve["knot_points"]=[[t,price*math.exp(log_move)] for t,price in curve["knot_points"]]
    for surface in snapshot["surfaces"]:
        surface["vols_row_major"]=[value+volatility for value in surface["vols_row_major"]]
        assert min(surface["vols_row_major"])>0
    for ticker,price in snapshot["prices"].items():
        if "price" in price:
            price["price"]["amount"]=str(float(price["price"]["amount"])*math.exp(equity))
        elif ticker.endswith("-VOL"):
            price["unitless"]+=volatility
    for quote in snapshot["fx"]["quotes"]:
        if quote[:2]==["EUR","USD"]:quote[2]*=math.exp(fx_move)
    scenario_market=MarketContext.from_json(json.dumps(snapshot))
    native_value=value_portfolio(book,scenario_market,metrics=[]).total_value
    tranche_value=0.
    if holding:
        scenario_deal=deepcopy(holding["deal"])
        spec=scenario_deal["instrument"]["spec"]
        # The zero-CDR baseline has a boundary: worsening systemic credit adds
        # annual collateral CDR; favorable shocks leave zero at zero. This mapping
        # is an explicit synthetic assumption, not an estimated pricing PD.
        cdr=spec["default_spec"]["cdr"]+.10*max(hazard,0.)
        spec["default_spec"]["cdr"]=cdr
        spec["default_assumptions"]["base_cdr_annual"]=cdr
        tranche_value=holding["quantity"]*structured_credit_tranche_metrics(
            json.dumps(scenario_deal),"BBB",scenario_market,AS_OF).pv
    return native_value,tranche_value

zero_native,zero_bbb=capstone_scenario_value(np.zeros(8))
assert abs(zero_native+zero_bbb-combined_base)<.01
scenario_values=np.array([capstone_scenario_value(shock) for shock in factor_panel])
scenario_pnl=scenario_values.sum(axis=1)-combined_base
scenario_returns=scenario_pnl/combined_base
assert np.isfinite(scenario_pnl).all() and len(scenario_pnl)==504
daily_risk=Performance.from_returns(pd.DataFrame({"held_book":scenario_returns},
    index=pd.bdate_range("2023-02-08",periods=504)),frequency="daily")
confidence=.95
q=float(daily_risk.value_at_risk(confidence).iloc[0])
tail=float(daily_risk.expected_shortfall(confidence).iloc[0])
manual_q=float(np.quantile(scenario_returns,1-confidence,method="linear"))
manual_tail=float(scenario_returns[scenario_returns<=manual_q].mean())
assert abs(q-manual_q)<1e-12 and abs(tail-manual_tail)<1e-12
bootstrap_rng=np.random.default_rng(2025011595)
bootstrap_risk=[]
for _ in range(500):
    sample=bootstrap_rng.choice(scenario_returns,size=len(scenario_returns),replace=True)
    cutoff=np.quantile(sample,1-confidence,method="linear")
    bootstrap_risk.append([-cutoff*combined_base,-sample[sample<=cutoff].mean()*combined_base])
bootstrap_interval=np.quantile(np.array(bootstrap_risk),[.025,.975],axis=0)
tail_risk_table=pd.DataFrame([{"confidence":confidence,"scenario_count":504,
    "VaR_loss_USD":-q*combined_base,"ES_loss_USD":-tail*combined_base,
    "tail_observations":int((scenario_returns<=manual_q).sum()),
    "VaR_bootstrap_p025":bootstrap_interval[0,0],"VaR_bootstrap_p975":bootstrap_interval[1,0],
    "ES_bootstrap_p025":bootstrap_interval[0,1],"ES_bootstrap_p975":bootstrap_interval[1,1],
    "mean_scenario_PnL_USD":scenario_pnl.mean(),"priced_positions":len(book.position_ids)+int(holding is not None)}])

fit=daily_risk.multi_factor_greeks(0,factor_panel.T.tolist())
design=np.column_stack([np.ones(len(factor_panel)),factor_panel])
ols=np.linalg.lstsq(design,scenario_returns,rcond=None)[0]
assert np.allclose(fit.betas,ols[1:],atol=1e-9)
residual=scenario_returns-design@ols
assert np.allclose(design.T@residual,0.,atol=1e-10)
covariance=np.cov(factor_panel,rowvar=False,ddof=1)
contributions=ols[1:]*(covariance@ols[1:])
residual_variance=residual.var(ddof=1)
total_variance=scenario_returns.var(ddof=1)
assert abs(contributions.sum()+residual_variance-total_variance)<1e-12
factor_risk_table=pd.DataFrame({"factor":factor_names,"return_loading_per_factor_unit":fit.betas,
    "USD_loading_per_factor_unit":np.array(fit.betas)*combined_base,
    "daily_return_variance_contribution":contributions})
factor_risk_table.loc[len(factor_risk_table)]={"factor":"nonlinear_residual",
    "return_loading_per_factor_unit":0.,"USD_loading_per_factor_unit":0.,
    "daily_return_variance_contribution":residual_variance}
print(tail_risk_table.to_string(index=False));print(factor_risk_table.to_string(index=False))
print({"factor_R_squared":fit.r_squared,"variance_reconciliation_error":contributions.sum()+residual_variance-total_variance})
print("These are 504 invented one-day market shocks with no time roll, flows or carry; date labels only satisfy the time-series API. Full native repricing preserves option nonlinearity and reprices credit BBB once per scenario. The selected parallel/co-moving factors omit tenor shape, subcurve and issuer basis, volatility smile changes, correlation jumps, liquidity and default jumps. VaR/ES describe this scenario model, not a complete regulatory or empirically calibrated risk estimate. ES follows the library's closed lower-tail sample convention; loss signs are displayed as positive USD. The 500-resample percentile intervals diagnose finite-sample uncertainty under this empirical scenario distribution, not model uncertainty or a guaranteed coverage bound. Factor coefficients are fitted conditional loadings, not exact instrument Greeks; small irrelevant-factor estimates and negative covariance contributions can occur.")
```

## 4. Stress — reprice the same contracts

Recalibrate common USD rates +150 bp and credit par spreads +100 bp, reduce SPX by 20%, and raise equity volatility. The credit variant also widens the convertible risky discount curve and stresses actual CLO collateral default/recovery assumptions. The volatility variant reprices the delivery bond and both commodity legs. All components, including BBB, are independently repriced; the combined P&L must reconcile.

```python
stress=tracks.build_market(TRACK)
calibration=tracks.common.calibration_envelope("base")
for quote in calibration["market_data"]:
    if quote["kind"]=="rate_quote":quote["rate"]+=.015
    elif quote["kind"]=="cds_quote":quote["spread_bp"]+=100.0
calibrated=calibrate(calibration).market
stress.insert(calibrated.get_discount("USD-OIS"))
stress.insert(calibrated.get_hazard("ACME-HZD"))
stress.insert_price("SPX-SPOT",4160.,"USD")
stress.insert(VolSurface("SPX-VOL",[.25,.5,1.,2.],[4000.,5200.,6400.],[[.35,.30,.29]]*4))
if TRACK=="credit":
    index_calibration=tracks.credit_calibration_envelope()
    for quote in index_calibration["market_data"]:
        if quote["kind"]=="rate_quote":quote["rate"]+=.015
        elif quote["kind"]=="cds_quote":quote["spread_bp"]+=100.0
    stress.insert(calibrate(index_calibration).market.get_hazard("CDX-HAZ"))
    stress.insert(DiscountCurve("CONVERT-RISKY",AS_OF,[(t,math.exp(-.095*t)) for t in [0.,1.,5.,10.]],day_count="act_365f"))
    stress.insert_price("CONVERT",80.,"USD")
    stressed_deal=deepcopy(holding["deal"])
    stressed_deal["instrument"]["spec"].update({"default_spec":{"cdr":.05,"curve":None},
        "recovery_spec":{"rate":.25,"recovery_lag":6}})
    stressed_deal["instrument"]["spec"]["default_assumptions"].update({"base_cdr_annual":.05,"base_recovery_rate":.25})
    bbb_stress=structured_credit_tranche_metrics(json.dumps(stressed_deal),"BBB",stress,AS_OF).pv*holding["quantity"]
else:
    stress.insert(DiscountCurve("USD-TREASURY",AS_OF,[(t,math.exp(-.055*t)) for t in [0.,1.,5.,10.,20.]],day_count="act_365f"))
    stress.insert_price("SPX",4160.,"USD")
    strikes=[5200*math.exp(-3+i*.05) for i in range(121)]
    stress.insert(VolSurface("SPX_VOL",[.25,.5,1.,2.],strikes,[[.30]*len(strikes)]*4))
    for ticker,levels,sigma in [("WTI",[75.,74.,72.,70.],.35),("BRENT",[78.,77.,75.,73.],.33)]:
        stress.insert(PriceCurve(f"{ticker}-FORWARD",AS_OF,list(zip([0.,.5,1.,2.],[.9*x for x in levels]))))
        stress.insert(VolSurface(f"{ticker}-VOL",[.25,.5,1.,2.],[20.,75.,150.],[[sigma]*3]*4))
    bbb_stress=0.0
stressed=value_portfolio(book,stress,metrics=[])
stress_wire=json.loads(stressed.to_json())
scenario=pd.DataFrame([{"position":iid,"base_usd":float(row["value_base"]["amount"]),
    "stress_usd":float(stress_wire["position_values"][iid]["value_base"]["amount"])} for iid,row in wire["position_values"].items()])
if holding:scenario.loc[len(scenario)]={"position":"CLO-BBB","base_usd":bbb_pv,"stress_usd":bbb_stress}
scenario["pnl_usd"]=scenario.stress_usd-scenario.base_usd
combined_stress=stressed.total_value+bbb_stress
assert abs(scenario.pnl_usd.sum()-(combined_stress-combined_base))<.01
print(scenario)
```

```python
from finstack_quant.models import bs_price
variance=tracks.variance_inputs()["SPX-VARIANCE"]
strike_grid=np.array([5200*math.exp(-3+i*.05) for i in range(121)])
width=np.empty(121);width[0]=(strike_grid[1]-strike_grid[0])/2;width[-1]=(strike_grid[-1]-strike_grid[-2])/2
width[1:-1]=(strike_grid[2:]-strike_grid[:-2])/2
base_df=market.get_discount("USD-OIS").df(1.)
base_forward=5200/base_df*math.exp(-.015)
cut=int(np.flatnonzero(strike_grid<=base_forward)[-1]);k0=strike_grid[cut]
replica_rows=[]
for label,m,s,vol in [("base",market,5200.,.20),("joint_stress",stress,4160.,.30)]:
    discount=m.get_discount("USD-OIS").df(1.);rate=-math.log(discount);forward=s/discount*math.exp(-.015)
    options=[]
    for i,k in enumerate(strike_grid):
        call=bs_price(s,float(k),rate,.015,vol,1.,True);put=bs_price(s,float(k),rate,.015,vol,1.,False)
        options.append(put if i<cut else call if i>cut else .5*(put+call))
    option_strip_pv=2*1_000_000*float(np.sum(width*np.array(options)/strike_grid**2))
    forward_cash_correction=-1_000_000*discount*(2*((forward/k0-1)-math.log(forward/k0))+.04)
    replication=option_strip_pv+forward_cash_correction
    native=price_instrument(json.dumps(variance),m,AS_OF,metrics=["variance_expected"]).value.amount
    replica_rows.append({"case":label,"fixed_option_inventory_PV":option_strip_pv,"forward_cash_correction":forward_cash_correction,
        "replication_PV":replication,"native_variance_PV":native,"finite_strike_residual":replication-native})
assert abs(replica_rows[0]["finite_strike_residual"])<.01
# Fixed finite strikes introduce a quantified quadrature residual when the forward moves across the grid.
assert abs(replica_rows[1]["finite_strike_residual"])<1_000
replication_table=pd.DataFrame(replica_rows)
replication_pnl=replication_table.iloc[1][["fixed_option_inventory_PV","forward_cash_correction","replication_PV","native_variance_PV"]]-replication_table.iloc[0][["fixed_option_inventory_PV","forward_cash_correction","replication_PV","native_variance_PV"]]
assert abs(replication_pnl.fixed_option_inventory_PV+replication_pnl.forward_cash_correction-replication_pnl.replication_PV)<.01
print(replication_table.to_string(index=False));print(replication_pnl)
print("The option inventory stays fixed; the forward/cash term represents the log-contract correction/dynamic-underlying component. It is not an unfinanced options-only hedge. The finite-grid residual is reported, not hidden.")
```

```python
from _shared.borrower_model import PERIODS
from finstack_quant.statements import ModelBuilder,ForecastSpec
borrower_spec=borrower_model();historical=Evaluator().evaluate(borrower_spec)
quarter_rows=[]
for scenario_name in ["base","downside"]:
    assumptions=borrower_inputs["scenarios"][scenario_name]
    builder=ModelBuilder.from_spec(borrower_spec)
    for node in ["revenue","cogs","opex"]:
        builder.forecast(node,ForecastSpec.growth(assumptions[node+"_growth_per_quarter"]))
    builder=builder.mixed("interest_rate").values([(q,historical.get("interest_rate",q)) for q in PERIODS[:4]]).formula(str(assumptions["annual_interest_rate"])).build()
    forecast=Evaluator().evaluate(builder.build())
    engine=CovenantEngine.from_specs([
        CovenantSpec(Covenant(CovenantType.max_net_debt_to_ebitda(4.5),"3M","Net leverage"),"net_leverage"),
        CovenantSpec(Covenant(CovenantType.min_dscr(1.1),"3M","DSCR"),"dscr")])
    for quarter,day in zip(PERIODS[4:],["2025-03-31","2025-06-30","2025-09-30","2025-12-31"]):
        assert abs(forecast.get("balance_check",quarter))<.01
        reports=engine.evaluate({"net_leverage":forecast.get("net_leverage",quarter),"dscr":forecast.get("dscr",quarter)},day)
        for name,out in reports.items():
            headroom=out.threshold-out.actual_value if name=="Net leverage" else out.actual_value-out.threshold
            assert abs(out.headroom-headroom/out.threshold)<1e-12
            quarter_rows.append({"scenario":scenario_name,"quarter":str(quarter),"covenant":name,
                "actual":out.actual_value,"threshold":out.threshold,"headroom":headroom,
                "native_headroom_fraction":out.headroom,"passed":out.passed})
credit_table=pd.DataFrame(quarter_rows)
assert len(credit_table)==16
assert not credit_table[(credit_table.scenario=="base")&(credit_table.covenant=="DSCR")].passed.any()
print(credit_table.to_string(index=False))
print("Base DSCR fails all four forecast tests. Headroom is shown in ratio units beside the native fraction; annualized EBITDA is a current-quarter run rate.")
```

## 7. Three-period attribution — actual dated repricing

This independent capstone history contains three consecutive one-week periods, January 15–February 5, 2025. It precedes coupon payments, option expiry and physical delivery, and contains no external flows or trades from the shared event ledger. Part III’s separate three-quarter history proves external-flow handling. Dated markets use the shared history observations and current FX at every date. BBB is repriced on every date. For the UST future, the mark can be transferred into the daily-VM cash ledger and its reference price reset; the wealth identity is shown explicitly. The variance swap receives every shared weekday close observed up to each valuation date, so its accrued realized variance is priced from dated inputs without look-ahead.

```python
from finstack_quant.core.market_data import ScalarTimeSeries
dates=[AS_OF+timedelta(days=7*i) for i in range(4)]
assert not [event for event in history_events() if dates[0]<date.fromisoformat(event["date"])<=dates[-1]]
assert dates[-1]<date(2025,3,20)  # Before the first future contractual coupon in these books.
nav_rows=[]
sleeve_rows=[]
for day in dates:
    quotes=history_quotes(day)
    dated=tracks.build_market(TRACK,day)
    common_market=history_market(day)
    dated.insert(common_market.get_discount("USD-OIS"))
    dated.insert(common_market.get_hazard("ACME-HZD"))
    dated.insert_price("SPX-SPOT",quotes["spx"],"USD")
    fx=FxMatrix();fx.set_quote(Currency("EUR"),Currency("USD"),quotes["eur_usd"]);dated.insert_fx(fx)
    if TRACK=="credit":
        index_calibration=tracks.credit_calibration_envelope(day)
        for quote in index_calibration["market_data"]:
            if quote["kind"]=="rate_quote":quote["rate"]+=quotes["rate_shift"]
            elif quote["kind"]=="cds_quote":quote["spread_bp"]+=quotes["credit_shift_bp"]
        dated.insert(calibrate(index_calibration).market.get_hazard("CDX-HAZ"))
        dated.insert_price("CONVERT",100*quotes["spx"]/5200,"USD")
    else:
        dated.insert_price("SPX",quotes["spx"],"USD")
        # Supply each realized close available on this valuation date, without look-ahead.
        observed_days=[AS_OF+timedelta(days=i) for i in range((day-AS_OF).days+1)]
        closes=[(observed,history_quotes(observed)["spx"]) for observed in observed_days if observed.weekday()<5]
        dated.insert_series(ScalarTimeSeries("SPX-CLOSE",closes))
        assert closes[-1][0]==day and closes[0]==(AS_OF,5200.)
        dated.insert(DiscountCurve("USD-TREASURY",day,[(t,math.exp(-(.04+quotes["rate_shift"])*t)) for t in [0.,1.,5.,10.,20.]],day_count="act_365f"))
    dated_spec=tracks.book_spec(TRACK);dated_spec["as_of"]=day.isoformat()
    valuation=value_portfolio(Portfolio.from_spec(json.dumps(dated_spec)),dated,metrics=[])
    dated_wire=json.loads(valuation.to_json())
    by_position={iid:float(row["value_base"]["amount"]) for iid,row in dated_wire["position_values"].items()}
    bbb_dated=structured_credit_tranche_metrics(json.dumps(holding["deal"]),"BBB",dated,day).pv*holding["quantity"] if holding else 0.0
    if TRACK=="vol":
        future=deepcopy(tracks.vol_extension()["UST-FUTURE"])
        future_mark=price_instrument(json.dumps(future),dated,day,model="bond_future_clean_price_proxy",metrics=["futures_price"])
        cumulative_vm_cash=future_mark.value.amount
        future["instrument"]["spec"]["quoted_price"]=future_mark.metrics["futures_price"]
        reset_mark=price_instrument(json.dumps(future),dated,day,model="bond_future_clean_price_proxy").value.amount
        assert abs(reset_mark)<.01
        assert abs(valuation.total_value-by_position["UST-FUTURE"]+reset_mark+cumulative_vm_cash-valuation.total_value)<.01
    else:cumulative_vm_cash=0.0
    sleeves={"common":sum(value for iid,value in by_position.items() if iid in common_ids),
             "track_extension":sum(value for iid,value in by_position.items() if iid not in common_ids)}
    if holding:sleeves["CLO-BBB"]=bbb_dated
    total=valuation.total_value+bbb_dated
    assert abs(sum(sleeves.values())-total)<.01
    nav_rows.append({"date":day,"native_pv_usd":valuation.total_value,"BBB_pv_usd":bbb_dated,
                     "combined_nav_usd":total,"EUR_USD":quotes["eur_usd"],"cumulative_futures_VM_cash_usd":cumulative_vm_cash})
    sleeve_rows.append(sleeves)
nav=pd.DataFrame(nav_rows)
print(nav)
```

```python
from finstack_quant.core.market_data import ScalarTimeSeries
# Use the actual final history date/market and only observations known at that date.
end_date=dates[-1];elapsed=(end_date-AS_OF).days/365;remaining=1-elapsed
observed=[(AS_OF+timedelta(days=i),history_quotes(AS_OF+timedelta(days=i))["spx"]) for i in range((end_date-AS_OF).days+1) if (AS_OF+timedelta(days=i)).weekday()<5]
dated.insert_series(ScalarTimeSeries("SPX-CLOSE",observed))
full=price_instrument(json.dumps(variance),dated,end_date,metrics=["variance_expected"])
forward_only=deepcopy(variance);forward_only["instrument"]["spec"]["start_date"]=end_date.isoformat()
forward_result=price_instrument(json.dumps(forward_only),dated,end_date,metrics=["variance_expected"])
initial=price_instrument(json.dumps(variance),market,AS_OF,metrics=["variance_expected"])
accrued=float(np.square(np.diff(np.log([close for _,close in observed]))).sum())
v0=initial.metrics["variance_expected"];v1=forward_result.metrics["variance_expected"]
df1=dated.get_discount("USD-OIS").df(remaining);df0=market.get_discount("USD-OIS").df(1.)
assert abs(full.metrics["variance_expected"]-(accrued+remaining*v1))<1e-10
carry_pnl=1_000_000*df1*(remaining*v0-.04)-initial.value.amount
realized_pnl=1_000_000*df1*accrued
implied_pnl=1_000_000*df1*remaining*(v1-v0)
assert abs(carry_pnl+realized_pnl+implied_pnl-(full.value.amount-initial.value.amount))<.01
variance_attribution=pd.DataFrame([{"component":"elapsed accrual/financing at initial forward variance","PnL":carry_pnl},
    {"component":"realized squared returns","PnL":realized_pnl},{"component":"remaining implied forward variance","PnL":implied_pnl}])
print(variance_attribution.to_string(index=False));print({"total_native_variance_PnL":full.value.amount-initial.value.amount,"realized_variance_sum":accrued,"initial_forward_variance":v0,"ending_forward_variance":v1})
```

The benchmark is the disclosed constant-cash 4% annual simple-return comparator, with equal benchmark sleeve weights. Brinson selection/interaction compares each repriced sleeve against that comparator; allocation can be zero because benchmark sleeve returns are equal. Carino links all three periods to the compounded active return. The result is a short model-mark history, not observed fund performance or a statistical sample for risk estimation.

```python
periods=[]
returns=[]
for i in range(3):
    start_total=nav.iloc[i].combined_nav_usd
    benchmark_return=.04*(dates[i+1]-dates[i]).days/365
    sectors=[]
    for sleeve,start_value in sleeve_rows[i].items():
        assert start_value>0
        sleeve_return=(sleeve_rows[i+1][sleeve]-start_value)/start_value
        sectors.append({"sector":sleeve,"portfolio_weight":start_value/start_total,
            "benchmark_weight":1/len(sleeve_rows[i]),"portfolio_return":sleeve_return,"benchmark_return":benchmark_return})
    attribution=brinson_fachler(json.dumps(sectors))
    actual_return=nav.iloc[i+1].combined_nav_usd/start_total-1
    assert abs(attribution.portfolio_return-actual_return)<1e-12
    assert abs(attribution.total_allocation+attribution.total_selection+attribution.total_interaction-attribution.total_excess_return)<1e-12
    periods.append(sectors);returns.append(actual_return)
linked=carino_link(json.dumps(periods))
linked_effects=linked.linked_allocation+linked.linked_selection+linked.linked_interaction
active=linked.portfolio_return_compounded-linked.benchmark_return_compounded
assert abs(linked_effects-active)<1e-12
wealth_return=nav.iloc[-1].combined_nav_usd/nav.iloc[0].combined_nav_usd-1
assert abs(linked.portfolio_return_compounded-wealth_return)<1e-12
attribution_table=linked.to_dataframe()
print(attribution_table)
```
